## Image Geometric Transformation

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
%matplotlib inline


#Define the translate  function
def translate(img, x,y):
    (h,w)=img.shape[:2]
    
    #Define the translation matric
    M=np.float32([[1,0,x],[0,1,y]])
    
    # Use the OpenCV affine transl]formation function to perform the translation
    shifted=cv2.warpAffine(img,M,(w,h))
    #Retrun the transformed image.
    return shifted


#Load the image and display it.
im=cv2.imread('panda.jpg')
im_RBG=cv2.cvtColor(im,cv2.COLOR_BGR2RGB)
plt.figure(figsize=(8,8))
plt.subplot(2,2,1),plt.title('orginal'),plt.imshow(im_RBG)

#Move down by 500 pixels 
shifted_1=translate(im_RGB,0,500)
plt.subplot(2,2,2),plt.title('down move 500 pixels'), plt.imshow(shifted_1)
# Move left 1000 pixels
shifted_2=translate(im_RBG, -1000,0)
plt.subplot(2,2,3), plt.title('left move 10 000 pixels'),plt.imshow(shifted_2)

#Move right by 500 pixels and down by 1,000 pixels
shifted_3=translate(im_RBG, 500,1000)
plt.subplot(2,2,4), plt.title('right move 500, down move 1000 pixels'), plt.imshow(shifted_3)
```

## Rotation

```python
#Define Rotate function
def rotate(img, angle, center=None, scale=1.0):
    (h,w)=img.shape[:2]
    if center is None:
        center=(w/2, h/2)
        
    M=cv2.getRotationMatrix2D(center , angle, scale)
    rotated=cv2.warpAffine(img, M, (w,h))
    # Return Rotated image
    return rotated

im=cv2.imread("panda.jpg")
im_RBG=cv2.cvtColor(im, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(8,8))
plt.subplot(2,2,1), plt.title('Original'), plt.imshow(im_RBG)

# Perform rotation on the original image
# counter clockwise rotaiton 45 degrees
rotated_45=rotate(im_RBG, 45)
plt.subplot(2,2,2), plt.title('rotate 45 degree  counter-clockwise'), plt.imshow(rotated_45)

# Clockwise by 20 degrees
rotated_minus20=rotate(im_RBG, -20)
plt.subplot(2,2,3), plt.title('rotate 20 degree  counter-clockwise'), plt.imshow(rotated_minus20)
# counter clockwise rotaiton 45 degrees
rotated_90=rotate(im_RBG, 90)
plt.subplot(2,2,2), plt.title('rotate 90 degree  counter-clockwise'), plt.imshow(rotated_90)
```

# Imagegrayscale Transformation

```python
# Define the linear grayscale transformation function
def linear_trans(img, k, b=0):
    #Calcualte the mapping table
    trans_list=[(np.float32(x)*k+b) for x in range(256)]
    # Convert the list to the array
    trans_table=np.array(trans_list)
    # Asjust the value
    trans_table[trans_table>255]=255
    trans_table[trans_table<0]=0
    trans_table=np.round(trans_table).astype(np.uint8)
    # Use OpenCV's look-up
    result = cv2.LUT(img, trans_table)
    
    return result
    
im=cv2.imread(r'panda.jpg',0)
plt.figure(figsize=(8,8))
plt.subplot(2,2,1), plt.title('Original'), plt.imshow(im,'gray')

#Inversion 
im_inversion=linear_trans(im, -1,255)
plt.subplot(2,2,2), plt.title('reversed'), plt.imshow(im_inversion, 'gray')
#Grayscale stretching
im_stretch=linear_trans(im, 1.2)
plt.subplot(2,2,3), plt.title('stretch'), plt.imshow(im_stretch, 'gray')

#Grayscale compression
im_compress=linear_trans(im, 0.4)
plt.subplot(2,2,4), plt.title('compress'), plt.imshow(im_compress, 'gray')
    
    
```

## Grammar transformation

```python
def gamma_trans(img, gamma):
    #Normalize
    gamma_list=[np.power(x/255.0, gamma)*255.0 for x in range(256)]
    gamma_table=np.round(np.array(gamma_list)).astype(np.uint8)
    
    return cv2.LUT(img, gamma_table)

    
im=cv2.imread(r'panda.jpg',0)
plt.figure(figsize=(8,8))
plt.subplot(1,3,1), plt.title('Original'), plt.imshow(im,'gray')

#Use gamma value 
im_gamma0=gamma_trans(im, 0.5)
plt.subplot(1,3,2), plt.title('Gamma=0.5'), plt.imshow(im_gamma0, 'gray')
#Grayscale stretching
im_gamma2=gamma_trans(im, 2)
plt.subplot(1,3,3), plt.title('Gamma=2'), plt.imshow(im_gamma2, 'gray')
```

## Histogram equalization

```python
im=cv2.imread(r'panda.jpg',0)
plt.figure(figsize=(8,8))
plt.subplot(2,2,1), plt.title('Original'), plt.imshow(im,'gray')

# Call OpenCV's histogram equalizer API
im_equal1=cv2.equalizeHist(im)
plt.subplot(2,2,3),plt.imshow(im_equal1 ,'gray')

#Display the histogram of the original image
plt.subplot(2,2,2)
plt.hist(im.ravel(), 256, [0,256], label='org')
plt.legend()

# Display the histogram of the equalized image
plt.subplot(2,2,4)
plt.hist(im_equal1.ravel(), 256,[0,256], label='equalize')
plt.legend()
plt.show()
```

## Image Binarization

```python
img=cv2.imread(r'panda.jpg',0)
# Simple threshold segmentation
ret1, th1=cv2.threshold(img,127, 255, cv2.THRESH_BINARY)
# Otsu filtering
ret2, th2=cv2.threshold(img, 0, 255, cv2.THRESH_OTSU)
print("Simple threshold vallue:", ret1)
print("Optimal threshold value:", ret2)
plt.figure(figsize=(8,8))

plt.subplot(2,2,1), plt.imshow(img, 'gray')
plt.subplot(2,2,2), plt.hist(img.ravel(), 256)
plt.subplot(2,2,3), plt.imshow(th1, 'gray')
plt.subplot(2,2,4), plt.imshow(th2, 'gray')
```

## Image Morphological Processing

```python
img=cv2.imread('panda.jpg',0)
_,bin_img=cv2.threshold(img,0,255, cv2.THRESH_OTSU)
plt.imshow(bin_img, 'gray')
```

```python
# Define structural elements
element=cv2.getStructuringElement(cv2.MORPH_CROSS,(5,5))
plt.imshow(element, 'gray')
```

## Erosion

```python
#Erode the image

eroded=cv2.erode(bin_img, element)
#Display the eroded image
plt.imshow(eroded, 'gray')
# Original image
plt.figure()
plt.imshow(bin_img, 'gray')
```

## Dilation

```python
# Dilate the image
dilated=cv2.dilate(bin_img, element)
# Display the dilate image
plt.imshow(dilated, 'gray')
#Original image
plt.figure()
plt.imshow(bin_img, 'gray')
```

# Image Filtering

### Median Filtering

```python
im=cv2.imread('panda.jpg')
im=cv2.cvtColor(im, cv2.COLOR_BGR2RGB)
plt.imshow(im)
# Call OpenCV's median blur API
im_medianblur=cv2.medianBlur(im,5)

plt.figure()
plt.imshow(im_medianblur)
plt.show()
```

## Mean Filtering

```python
im=cv2.imread('panda.jpg')
im=cv2.cvtColor(im, cv2.COLOR_BGR2RGB)
plt.imshow(im)
# Call OpenCV's median blur API
im_meanblur=cv2.blur(im,(3,3))

plt.figure()
plt.imshow(im_meanblur)
plt.show()
```

## Gaussian Filtering

```python
im=cv2.imread('panda.jpg')
im=cv2.cvtColor(im, cv2.COLOR_BGR2RGB)
plt.imshow(im)
# Call OpenCV's median blur API
im_gaussianblur=cv2.GaussianBlur(im,(5,5),0)

plt.figure()
plt.imshow(im_gaussianblur)
plt.show()
```

## Image sharpening

```python
im=cv2.imread('panda.jpg')
im=cv2.cvtColor(im, cv2.COLOR_BGR2RGB)
plt.imshow(im)

#Sharpening operator
sharpen_1=np.array([
    [-1,-1,-1],
    [-1,9,-1],
    [-1,-1,-1]
])
# Use filter2D for filtering operations
im_sharpen1=cv2.filter2D(im,-1,sharpen_1)
plt.figure()
plt.imshow(im_sharpen1)
# Sharpening operator 2
sharpen_2=np.array([
    [0,-1,0],
    [-1,8,-1],
    [0,1,0]
])/4.0

# Use filter2D for filtering operations
im_sharpen2=cv2.filter2D(im, -1, sharpen_2)
plt.figure()
plt.imshow(im_sharpen2)
```
